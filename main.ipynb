```python
# 19. Outlier Detection for BMI and Charges Using IQR Method
for col in ["bmi", "charges"]:

    Q1, Q3 = df[col].quantile(0.25), df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower, upper = Q1 - 1.5 * IQR, Q3 + 1.5 * IQR

    n_out = ((df[col] < lower) | (df[col] > upper)).sum()

    print(f"{col}: bounds = ({lower:.1f}, {upper:.1f}), outliers = {n_out} ({n_out/len(df) * 100: .1f}%)")
```

```python
# 20. Medical Charges Summary by Smoking Status
df.groupby("smoker")["charges"].agg(["mean", "median", "std", "count"])
```

```python
# 21. Define BMI Category Binning Function
def bmi_category(bmi):
    if bmi < 18.5:
        return "underweight"
    elif bmi < 25:
        return "normal"
    elif bmi < 30:
        return "overweight"
    else:
        return "obese"
```

```python
# 22. Apply BMI Categorization to Create Feature (bmi_category)
df["bmi_category"] = df["bmi"].apply(bmi_category)
```

```python
# 23. Preview Dataset with New BMI Category Feature
df.head()
```

```python
# 25. Average Medical Charges Across BMI Categories
df.groupby("bmi_category")["charges"].mean().sort_values().round(2)
```

```python
# 26. Log Transformation of Target Variable (log_charges)
df["log_charges"] = np.log1p(df["charges"])
```

```python
# 27. Preview Dataset with Log-Transformed Charges
df.head()
```

```python
# 28. Skewness of Original Charges
df["charges"].skew()
```

```python
# 29. Skewness of Log-Transformed Charges
df["log_charges"].skew()
```

```python
# 30. One-Hot Encode Categorical Features
df_model = pd.get_dummies(
    df,
    columns=["sex", "smoker", "region", "bmi_category"],
    drop_first = True
)

print(df_model.columns.to_list())
```

```python
# 31. Preview Encoded Dataset for Modeling
df_model.head()
```

```python
# 32. Separate Features (X) and Target Variable (y)
feature_cols = [c for c in df_model.columns if c not in ["charges", "log_charges"]]

X = df_model[feature_cols]
y = df_model["log_charges"]
```

```python
# 33. Split Data into Training and Testing Sets
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.2)
```

```python
y_test_real = np.expm1(y_test)

candidate_models = {
    "Linear Regression": LinearRegression(),
    "Random Forest": RandomForestRegressor(n_estimators= 300, max_depth=6, n_jobs= -1),
    "Gradient Boosting": GradientBoostingRegressor(n_estimators= 300, max_depth= 3, learning_rate= 0.05)
}

trained_models = {}
metrics_rows = []

for name, model in candidate_models.items():

    model.fit(X_train, y_train)
    y_pred_log = model.predict(X_test)
    y_pred = np.expm1(y_pred_log)

    r2 = r2_score(y_test_real, y_pred)
    mae = mean_absolute_error(y_test_real, y_pred)
    rmse = np.sqrt(mean_squared_error(y_test_real, y_pred))

    trained_models[name] = model
    metrics_rows.append({"Model": name, "R2": r2, "MAE": mae, "RMSE": rmse})

    print(f"{name} --> R2: {r2:.4f} -- MAE: {mae:.2f} -- RMSE: {rmse:.2f}")
```

```python
results = pd.DataFrame(metrics_rows).sort_values("R2", ascending=False).reset_index(drop=True)
```

```python
results.round(4)
```

```python
best_model_name = results.loc[0, "Model"]
best_model = trained_models[best_model_name]
```
